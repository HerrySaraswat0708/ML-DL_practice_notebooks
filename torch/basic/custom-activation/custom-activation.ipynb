# Problem: Write a Custom Activation Function

### Problem Statement
You are tasked with implementing a **custom activation function** in PyTorch that computes the following operation:

$ \text{activation}(x) = \tanh(x) + x $

Once implemented, this custom activation function will be used in a simple linear regression model.

### Requirements
1. **Custom Activation Function**:
   - Implement a class `CustomActivationModel` inheriting from `torch.nn.Module`.
   - Define the `forward` method to compute the activation function \( \text{tanh}(x) + x \).

2. **Integration with Linear Regression**:
   - Use the custom activation function in a simple linear regression model.
   - The model should include:
     - A single linear layer.
     - The custom activation function applied to the output of the linear layer.

### Constraints
- The custom activation function should not have any learnable parameters.
- Ensure compatibility with PyTorch tensors for forward pass computations.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Generate synthetic data
torch.manual_seed(42)
X = torch.rand(100, 1) * 10  # 100 data points between 0 and 10
y = 2 * X + 3 + torch.randn(100, 1)  # Linear relationship with noise
```

```python
# Define the Linear Regression Model
class CustomActivationModel(nn.Module):
    def __init__(self):
        super(CustomActivationModel, self).__init__()
        self.linear = nn.Linear(1, 1)  # Single input and single output

    # TODO: Implement the forward pass
    def custom_activation(self, x):
        tanh = nn.Tanh()
        tanh_x = tanh(x)
        return tanh_x + x

    # TODO: Implement the forward pass
    def forward(self, x):
        return self.custom_activation(self.linear(x))

# Initialize the model, loss function, and optimizer
model = CustomActivationModel()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Training loop
epochs = 1000
for epoch in range(epochs):
    # Forward pass
    predictions = model(X)
    loss = criterion(predictions, y)

    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # Log progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
# Display the learned parameters
[w, b] = model.linear.parameters()
print(f"Learned weight: {w.item():.4f}, Learned bias: {b.item():.4f}")

# Testing on new data
X_test = torch.tensor([[4.0], [7.0]])
with torch.no_grad():
    predictions = model(X_test)
    print(f"Predictions for {X_test.tolist()}: {predictions.tolist()}")
```

```python
In a neural network, especially in a linear regression model before an activation function, the input `x` first passes through a linear layer (`self.linear(x)`) to perform a linear transformation. This transformation typically involves a matrix multiplication (weights) and an addition (bias): `y_linear = W * x + b`.

This linear step is crucial for several reasons:

1.  **Feature Combination and Transformation**: The linear layer combines the input features in a weighted sum. It learns which features are important and how they contribute to the output. Without this, the model would not be able to learn complex relationships in the data.
2.  **Learning Parameters**: The `W` (weights) and `b` (bias) are the learnable parameters of the model. The linear layer is where the model learns to map the input `x` to an appropriate space before the activation function is applied. The activation function itself usually has no learnable parameters, as is the case with your `custom_activation`.
3.  **Foundation for Non-Linearity**: The activation function (like your `tanh(x) + x`) introduces non-linearity into the model. If you only applied an activation function directly to `x` without a preceding linear layer, you would effectively be applying a non-linear transformation to the raw input features without learning any relationships or transformations between them. The linear layer first projects the input into a new space where the non-linear activation can then operate effectively to model complex patterns.

In essence, the linear layer provides the learnable transformation, and the activation function adds the necessary non-linearity to allow the model to learn more complex functions than just simple linear mappings.
```
