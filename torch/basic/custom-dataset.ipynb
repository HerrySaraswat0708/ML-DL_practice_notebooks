# Problem: Write a custom Dataset and Dataloader to load from a CSV file

### Problem Statement
You are tasked with creating a **custom Dataset** and **Dataloader** in PyTorch to load data from a given `data.csv` file. The loaded data will be used to run a pre-implemented linear regression model.

### Requirements
1. **Dataset Class**:
   - Implement a class `CustomDataset` that:
     - Reads data from a provided `data.csv` file.
     - Stores the features (X) and target values (Y) separately.
     - Implements PyTorch's `__len__` and `__getitem__` methods for indexing.

2. **Dataloader**:
   - Use PyTorch's `DataLoader` to create an iterable for batch loading the dataset.
   - Support user-defined batch sizes and shuffling of the data.

```python
import torch
import pandas as pd

torch.manual_seed(42)
X = torch.rand(100, 1) * 10  # 100 data points between 0 and 10
y = 2 * X + 3 + torch.randn(100, 1)  # Linear relationship with noise

# Save the generated data to data.csv
data = torch.cat((X, y), dim=1)
df = pd.DataFrame(data.numpy(), columns=['X', 'y'])
df.to_csv('data.csv', index=False)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
from numpy import dtype
import torch
from torch.utils.data import Dataset, DataLoader
import pandas as pd

# TODO: Add the missing code
class LinearRegressionDataset(Dataset):
    def __init__(self, csv_path):
      df = pd.read_csv(csv_path)
      self.X = df['X'].values
      self.y = df['y'].values

    def __len__(self):
      return len(self.X)

# if we don't apply torch.tensor([]) then dataloader will group 32 values and and make a tensor of shape [32] whereas we
# nees (32,1) thus either apply this or so unsueeze(0/1) to add extra dimension
    def __getitem__(self,idx):
      # Ensure that X and y are returned as 1-element tensors to maintain shape (batch_size, 1)
      return torch.tensor([self.X[idx]]), torch.tensor([self.y[idx]])

# Example usage of the DataLoader
dataset = LinearRegressionDataset('data.csv')
# TODO: Add the missing code
batch_size = int(input('enter batch size :'))
dataloader = DataLoader(dataset,batch_size=batch_size,shuffle=True)
```

```python
# Define the Linear Regression Model
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.linear = nn.Linear(1, 1)  # Single input and single output

    def forward(self, x):
        print(x.shape)
        print(self.linear.weight.shape)
        return self.linear(x)

# Initialize the model, loss function, and optimizer
model = LinearRegressionModel()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Training loop
epochs = 1000
for epoch in range(epochs):
    for batch_X, batch_y in dataloader:
        # Forward pass
        # Cast batch_X and batch_y to torch.float32 and add a dimension to match model's expected input shape
        predictions = model(batch_X.float())
        loss = criterion(predictions, batch_y.float())

        # Backward pass and optimization
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    # Log progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
# Display the learned parameters
[w, b] = model.linear.parameters()
print(f"Learned weight: {w.item():.4f}, Learned bias: {b.item():.4f}")

# Testing on new data
X_test = torch.tensor([[4.0], [7.0]])
with torch.no_grad():
    predictions = model(X_test)
    print(f"Predictions for {X_test.tolist()}: {predictions.tolist()}")
```
